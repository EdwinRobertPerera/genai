```python
import pandas as pd
import matplotlib.pyplot as plt
#Task 1: pandas series basics
marks =[78, 85, 90, 66, 72]
s1 = pd.Series(marks)
print('Mark Index:',s1.index,"\nMark Values:",s1.values,"\nMark Datatype:",s1.dtype)
print ("\nfirst element of marks series:",s1[0])
print ("\nLast two elements of marks series:",s1[-1:-3:-1])

#Task 2: Mathematical operations on series
s2=s1+5
print(s2)
s3=s1-2
print(s3)
s4=s1*1.05
print(s4)
s5=s1/2
print(s5)
#Task 3: Python Functionalities on Series
print ("\nMaximum marks:", s1.max())
print("\nMinimum marks:", s1.min())
print("\nSum of marks:", s1.sum())
print("\nMean of marks:", s1.mean())
c=0
print(s1.apply(lambda x: print("pass")if x>=70 else print("Fail")))
print(len(s1 [s1>=70]))

#Task 4 : Important DataFrame Functions
student_details={'Name':['Amit','Neha','Rahul','Sneha','Pooja'], 'Marks':[78, 85, 90, 66, 72],'Subject':['Maths','Maths','Science','Science','Maths']}
students=pd.DataFrame(student_details)
print(students)
print(students.iloc[0:3])
print(students.iloc[-3:(len(students))])
print(students.shape)
print(students.columns)

#Task 5 Important Dataframe functions
print(students.info())
print(students.describe())
print(students.head())
print(students.tail())
students_sorted=students.sort_values(by="Marks", ascending=False)
print(students_sorted)
students_sorted=students_sorted.sort_index()
print(students_sorted)

#Task 6 Filtering and Conditional Selection
print(students[students['Marks']>75])
print(students[students["Subject"]=='Maths'])
print(students["Marks"].mean())
print(students[students['Marks']>(students['Marks'].mean())])
print(students[students['Marks']<70])

#Task 8 Pandas Plotting 
students["Marks"].plot(kind="line",color='blue')
plt.show()
students["Marks"].plot(kind="bar",color='blue')
plt.show()
students["Marks"].plot(kind="hist",bins=5)
plt.show()


#Task 9 Mini Use Case : Sales Data Analysis
sales={'Day':['Mon','Tue','Wed','Thur','Fri'],'Revenue':[1200,1500,900,2000,1800]}
sales=pd.DataFrame(sales)
print(sales['Revenue'].sum())
print(sales['Revenue'].mean())
print(sales['Revenue'].max())
print(sales[sales['Revenue']>sales['Revenue'].mean()])
sales.plot(y='Revenue',x='Day',kind="line",color='blue')
plt.show()
```

```python
# Load dataset
df = pd.read_csv('csv files/sales_data.csv')

# Convert date and extract year
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
df['Year'] = df['Sale_Date'].dt.year

# Total sales by year and sales rep
sales_by_year_rep = df.groupby(['Year', 'Sales_Rep'])['Sales_Amount'].sum().unstack(fill_value=0)

# Prepare plot
years = sales_by_year_rep.index.tolist()
sales_reps = sales_by_year_rep.columns.tolist()
x = range(len(years))
bar_width = 0.2

plt.figure(figsize=(12, 6))

for i, rep in enumerate(sales_reps):
    offset = (i - (len(sales_reps) - 1) / 2) * bar_width
    plt.bar([pos + offset for pos in x], sales_by_year_rep[rep].values, width=bar_width, label=rep)

# Labels and title
plt.xticks([pos for pos in x], years)
plt.title('Multiple Bar Chart: Sales by Year and Sales Rep')
plt.xlabel('Year')
plt.ylabel('Total Sales Amount')
plt.legend(title='Sales Rep')
plt.tight_layout()
plt.show()
```

```python
#Task 5: Stacked Bar Chart
# Group by Product Category and sum Sales_Amount and Unit_Cost
stacked_data = df.groupby('Product_Category')[['Sales_Amount', 'Unit_Cost']].sum()

plt.figure(figsize=(10, 6))
stacked_data.plot(kind='bar', stacked=True, figsize=(10, 6), color=['skyblue', 'orange'])
plt.title('Stacked Bar Chart: Sales Amount vs Unit Cost by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Amount')
plt.xticks(rotation=45)
plt.legend(title='Metrics', labels=['Sales Amount', 'Unit Cost'])
plt.tight_layout()
plt.show()
```

```python
#Task 6: Histogram - Sales Amount Distribution
plt.figure(figsize=(10, 6))
plt.hist(df['Sales_Amount'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram: Sales Amount Distribution')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#Task 7: Pie Chart - Product Category Market Share
category_sales = df.groupby('Product_Category')['Sales_Amount'].sum()

plt.figure(figsize=(10, 8))
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff99cc']
plt.pie(category_sales, labels=category_sales.index, autopct='%1.1f%%', 
        startangle=90, colors=colors)
plt.title('Pie Chart: Product Category Market Share')
plt.tight_layout()
plt.show()
```
